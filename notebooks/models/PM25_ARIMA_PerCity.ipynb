```python
import numpy as np
import pandas as pd

from statsmodels.tsa.arima.model import ARIMA
from statsmodels.tsa.stattools import adfuller
from statsmodels.graphics.tsaplots import plot_acf, plot_pacf
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("../../data/master/PM25_2020 cleaned Master_20_23.csv")
df["Date"] = pd.to_datetime(df["Date"])

required_cols = {"Date", "City", "PM25"}
missing = required_cols - set(df.columns)
if missing:
    raise ValueError(f"Missing columns: {missing}")

print(df.shape)
```

```python
results = []

for city in cities:
    print(f"\nProcessing {city}")

    # Prepare city series
    city_df = df[df["City"] == city].copy()
    city_df = city_df.sort_values("Date")
    city_df = city_df.drop_duplicates(subset=["Date"], keep="last")
    city_df = city_df.set_index("Date").asfreq("D")
    city_df["PM25"] = city_df["PM25"].interpolate(limit_direction="both")

    # Train / test split
    y_train = city_df.loc[city_df.index < cutoff, "PM25"]
    y_test = city_df.loc[city_df.index >= cutoff, "PM25"]

    print("Train:", len(y_train), "| Test:", len(y_test))

    # Log transform
    y_train_log = np.log1p(y_train)

    # Wider ARIMA grid
    orders = []
    for p in range(0, 4):
        for d in [0, 1]:
            for q in range(0, 4):
                orders.append((p, d, q))

    # Select best model by TEST RMSE
    best_rmse = float("inf")
    best_order = None
    best_model = None
    best_aic = None

    for order in orders:
        try:
            model = ARIMA(y_train_log, order=order)
            fitted = model.fit()

            y_pred_log = fitted.forecast(steps=len(y_test))
            y_pred = np.expm1(y_pred_log)
            y_pred = pd.Series(y_pred, index=y_test.index).clip(lower=0)

            rmse = np.sqrt(mean_squared_error(y_test, y_pred))

            if rmse < best_rmse:
                best_rmse = rmse
                best_order = order
                best_model = fitted
                best_aic = fitted.aic

        except Exception:
            continue

    if best_model is None:
        print(f"No ARIMA model worked for {city}")
        continue

    # Final forecast using best model
    y_pred_log = best_model.forecast(steps=len(y_test))
    y_pred = np.expm1(y_pred_log)
    y_pred = pd.Series(y_pred, index=y_test.index).clip(lower=0)

    # Metrics
    mae = mean_absolute_error(y_test, y_pred)
    rmse = np.sqrt(mean_squared_error(y_test, y_pred))
    r2 = r2_score(y_test, y_pred)

    results.append({
        "City": city,
        "Model": "ARIMA",
        "Order": str(best_order),
        "AIC": best_aic,
        "MAE": mae,
        "RMSE": rmse,
        "R2": r2
    })

    print("Best order:", best_order)
    print("MAE:", round(mae, 3))
    print("RMSE:", round(rmse, 3))
    print("R2:", round(r2, 3))

    # Plot
    plt.figure(figsize=(12, 5))
    plt.plot(y_test.index, y_test.values, label="Actual")
    plt.plot(y_pred.index, y_pred.values, label="ARIMA Forecast")
    plt.title(f"ARIMA Forecast - {city} | order={best_order}")
    plt.legend()
    plt.show()
```
